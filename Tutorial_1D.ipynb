```julia
function H₀(k, p) # Haldane
    k1, k2 = k
    t₁ = 1
    t₂, ϕ, m = p

    h0 = 2t₂ * cos(ϕ) * (cos(k1) + cos(k2) + cos(k1 + k2))
    hx = t₁ * (1 + cos(k1) + cos(k2))
    hy = t₁ * (-sin(k1) + sin(k2))
    hz = m - 2t₂ * sin(ϕ) * (sin(k1) + sin(k2) - sin(k1 + k2))

    s0 = [1 0; 0 1]
    sx = [0 1; 1 0]
    sy = [0 -im; im 0]
    sz = [1 0; 0 -1]

    h0 .* s0 .+ hx .* sx .+ hy .* sy .+ hz .* sz
end

H(k) = H₀(k, (1, π/3, 0.5))
showBand(H; value=false, disp=true)
```
